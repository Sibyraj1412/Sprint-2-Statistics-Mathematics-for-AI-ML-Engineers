# Sprint -2 Measures of Central Tendency 

**1. Mean**

**Definition**
The mean is what most people call the average. You add up all the numbers and divide by how many numbers there are.

**Formula**
x̄ = (sum of all values) / (number of values)

**Numerical Example**
Marks of 5 students: 45, 60, 55, 70, 50
Mean = (45+60+55+70+50) / 5 = 280 / 5 = 56

**Business Example**
A retail company looks at the average monthly revenue across its 12 stores to set a target for the year.

**AI/ML Example**
The mean is used to fill in missing numbers in a dataset, and it is also the basic idea behind common loss functions like Mean Squared Error.

**Interpretation**
The mean is easy to compute and understand, but it gets pulled around by extreme values. If one salesperson has a huge outlier month, it can make the average look better than what most people actually experienced.

```python
import numpy as np
import pandas as pd

marks = [45, 60, 55, 70, 50]
data = pd.Series(marks)

print("Mean (pandas):", data.mean())
print("Mean (numpy):", np.mean(marks))
```

**2. Median**

**Definition**
The median is the middle value once you line up all the numbers in order. If there is an even number of values, you take the average of the two in the middle.

**Formula**
If n is odd, median is the value at position (n+1)/2
If n is even, median is the average of the two values at positions n/2 and n/2 + 1

**Numerical Example**
Salaries in thousands: 25, 28, 30, 32, 120
Sorted, the middle value is 30, so median = 30
Notice the mean here would be 47, pulled up a lot by the 120 outlier.

**Business Example**
Real estate companies usually report median house prices in a city, not the mean, because a handful of very expensive homes would otherwise skew the number upward.

**AI/ML Example**
When filling in missing values for something like income or transaction amount, people often use the median instead of the mean, because these kinds of features tend to have extreme values that distort the mean.

**Interpretation**
The median is not affected by extreme values the way the mean is. It gives a better sense of what is typical when the data is skewed, like income or house prices.

```python
import pandas as pd

salaries = [25, 28, 30, 32, 120]
data = pd.Series(salaries)

print("Median:", data.median())
print("Mean:", data.mean())
```

**3. Mode**

**Definition**
The mode is simply the value that shows up most often. A dataset can have one mode, more than one, or none at all.

**Formula**
There is no algebraic formula. You count how often each value appears and pick the one with the highest count.

**Numerical Example**
Shoe sizes sold in a day: 7, 8, 8, 9, 8, 10, 7
Size 8 shows up 3 times, more than any other, so the mode is 8.

**Business Example**
A shoe store looks at which size sells the most and stocks more of that size, since it tells them what customers buy most often.

**AI/ML Example**
For categorical columns like city or payment method, missing values are usually filled with the mode, since mean and median do not make sense for non-numeric data.

**Interpretation**
The mode tells you what is most common or popular. It works well for categories, but if a dataset has several values tied for the most frequent, the mode alone will not tell the full story.

```python
import pandas as pd
from statistics import mode, multimode

shoe_sizes = [7, 8, 8, 9, 8, 10, 7]
data = pd.Series(shoe_sizes)

print("Mode:", data.mode().tolist())
print("Mode using statistics module:", mode(shoe_sizes))
print("All modes:", multimode(shoe_sizes))
```

**4. Weighted Mean**

**Definition**
A weighted mean is an average where some values count more than others. Instead of treating every number equally, you multiply each one by a weight first.

**Formula**
x̄w = (sum of weight times value) / (sum of weights)

**Numerical Example**
A student's final grade is based on assignments worth 30 percent with a score of 80, a midterm worth 30 percent with a score of 70, and a final exam worth 40 percent with a score of 90.
Weighted mean = (0.30×80 + 0.30×70 + 0.40×90) = 24 + 21 + 36 = 81

**Business Example**
An investor works out the weighted average return of a portfolio, where each stock's return is weighted by how much money is put into it.

**AI/ML Example**
Weighted averages show up in loss functions when one class matters more than another, like giving more penalty for missing a fraud case than for a false alarm.

**Interpretation**
The weighted mean is useful whenever some data points genuinely matter more than others, like exam weightage or money invested. Ignoring the weights and just averaging normally would give a misleading picture.

```python
import numpy as np

scores = [80, 70, 90]
weights = [0.30, 0.30, 0.40]

weighted_mean = np.average(scores, weights=weights)
print("Weighted Mean:", weighted_mean)
```

**5. Geometric Mean**

**Definition**
The geometric mean is used when values multiply together instead of add together, like growth rates over time. You multiply all the values and then take the nth root.

**Formula**
GM = nth root of (x1 × x2 × ... × xn)

**Numerical Example**
A company's growth rates over 3 years were 10 percent, 20 percent, and 5 percent, written as 1.10, 1.20, 1.05.
GM = (1.10 × 1.20 × 1.05)^(1/3) is about 1.1148, which means about 11.48 percent average yearly growth.

**Business Example**
Companies use the geometric mean to report their compound annual growth rate over several years, since the regular average would overstate the real growth.

**AI/ML Example**
Geometric mean shows up in evaluation metrics for imbalanced datasets, where you take the geometric mean of sensitivity and specificity to check the model is not just favoring the majority class.

**Interpretation**
The geometric mean is always equal to or smaller than the arithmetic mean. It is the right tool whenever you are averaging growth rates, returns, or anything that compounds over time.

```python
from scipy import stats

growth_rates = [1.10, 1.20, 1.05]

gm = stats.gmean(growth_rates)
print("Geometric Mean:", round(gm, 4))
print("Average annual growth rate:", round((gm - 1) * 100, 2), "percent")
```

**6. Harmonic Mean**

**Definition**
The harmonic mean is used for averaging rates, like speed. You take the reciprocal of each value, average those, and then take the reciprocal again.

**Formula**
HM = n / (sum of 1/x for each value)

**Numerical Example**
A car drives the same distance at three different speeds: 60 km/h, 40 km/h, and 30 km/h.
HM = 3 / (1/60 + 1/40 + 1/30) = 3 / 0.075 = 40 km/h average speed

**Business Example**
Investors sometimes use the harmonic mean to average price-to-earnings ratios across a portfolio, since it does not let unusually high ratios dominate the result the way a normal average would.

**AI/ML Example**
The F1 score, used to judge classification models, is the harmonic mean of precision and recall. It is chosen over a plain average because it punishes the model harder if either precision or recall is really low.

**Interpretation**
The harmonic mean is always the smallest of the three means. It is the right choice when you are dealing with rates or ratios, because it stops big values from making the average look better than it really is.

```python
from scipy import stats

speeds = [60, 40, 30]

hm = stats.hmean(speeds)
print("Harmonic Mean (average speed):", round(hm, 2), "km/h")
```
